```python
# Q1 - bonus marks with broadcasting
marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])

bonus = np.array([5, 2, 8])
marks_new = marks + bonus
print("After bonus:")
print(marks_new)

capped_count = np.sum(marks_new > 100)
print("Marks that had to be capped:", capped_count)

marks_new[marks_new > 100] = 100
print("After capping:")
print(marks_new)
```

```python
# Q4 - hstack vs column_stack
heights = np.array([172, 158, 181, 166])
weights = np.array([68, 52, 84, 59])

joined = np.hstack((heights, weights))
print("hstack shape:", joined.shape)  # (8,) - just one long row

table = np.column_stack((heights, weights))
print("Table shape:", table.shape)  # (4, 2)
print(table)
# hstack didnt give 2 columns because both are 1D so it just joins them end to end
```

```python
# Q5 - quarterly sales
sales = np.array([120, 135, 98, 160, 175, 190, 88, 92, 105, 210, 198, 230])
quarters = sales.reshape(4, 3)
print("Quarters:")
print(quarters)

avg_per_quarter = quarters.mean(axis=1)
print("Averages:", avg_per_quarter)

best_q = np.argmax(avg_per_quarter)
print(f"Best quarter: Q{best_q + 1} with average {avg_per_quarter[best_q]:.2f}")
```

```python
# Q8 - column-wise min-max normalization
X = np.array([[ 50, 1200, 3],
              [ 80, 1500, 9],
              [ 65,  900, 5],
              [ 95, 2000, 7]])

col_min = X.min(axis=0)
col_max = X.max(axis=0)
X_norm = (X - col_min) / (col_max - col_min)
print("Normalized:")
print(X_norm)

# verify
print("Column mins:", X_norm.min(axis=0))
print("Column maxs:", X_norm.max(axis=0))
```

```python
# Q9 - nested np.where for medals
scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])

medals = np.where(scores >= 90, 'Gold',
         np.where(scores >= 75, 'Silver',
         np.where(scores >= 60, 'Bronze', 'None')))

print("Medals:", medals)
for m in ['Gold', 'Silver', 'Bronze', 'None']:
    print(f"{m}: {np.sum(medals == m)}")
```

```python
# Q10 - np.where to find elements > 50
m = np.array([[10, 80, 45],
              [55, 20, 99],
              [70, 33, 12]])

rows, cols = np.where(m > 50)
print("Row indices:", rows)
print("Col indices:", cols)
print("Elements > 50:", m[rows, cols])

m_copy = m.copy()
m_copy[rows, cols] = 0
print("Modified copy:")
print(m_copy)
print("Original unchanged:")
print(m)
```

```python
# Q11 - stacking sensors
s1 = np.array([12, 15, 11, 19, 14])
s2 = np.array([22, 18, 25, 21, 20])
s3 = np.array([31, 35, 29, 33, 30])

row_stack = np.vstack((s1, s2, s3))  # each sensor is a row
col_stack = np.column_stack((s1, s2, s3))  # each sensor is a column

print("Row stack shape:", row_stack.shape)
print("Col stack shape:", col_stack.shape)
print("One is transpose of other:", np.array_equal(row_stack, col_stack.T))
```

```python
# Q13 - matrix multiplication
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8, 9],
              [1, 0, 2]])

r1 = A @ B.T
r2 = A.T @ B
print("A @ B.T shape:", r1.shape)  # (2,3) @ (3,2) = (2,2)
print(r1)
print("A.T @ B shape:", r2.shape)  # (3,2) @ (2,3) = (3,3)
print(r2)

# A @ B doesnt work because A is (2,3) and B is (2,3)
# for matrix multiplication inner dimensions must match
# 3 != 2 so it fails
```

```python
# Q14 - multi-step pipeline
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

# step 1 - column vectors
h_col = hours.reshape(-1, 1)
a_col = attendance.reshape(-1, 1)
print("Column vector shape:", h_col.shape)

# step 2 - two column table
table = np.hstack((h_col, a_col))
print("Table shape:", table.shape)

# step 3 - add total column
totals = table.sum(axis=1).reshape(-1, 1)
table3 = np.hstack((table, totals))
print("With totals shape:", table3.shape)

# step 4 - weighted score
w = np.array([0.5, 0.2, 0.3])
scores = table3 @ w
print("Scores shape:", scores.shape)
print("Scores:", scores)

# step 5 - above average = 1, else 0
avg_score = scores.mean()
labels = np.where(scores > avg_score, 1, 0)
print("Labels:", labels)
```

```python
# Q15 - min-max normalization
arr = np.array([10, 25, 50, 75, 100])
normalized = (arr - arr.min()) / (arr.max() - arr.min())
print(normalized)
```

```python
# Q16 - flatten vs ravel
m = np.array([[4, 8, 15],
              [16, 23, 42]])

flat = m.flatten()
rav = m.ravel()

flat[0] = 999
rav[0] = 999

print("Original matrix:")
print(m)
# flatten() returns a copy so changing it doesnt affect original
# ravel() returns a view so changing it DOES change the original
# thats why m[0][0] is now 999
```

```python
# Q17 - weekly temperatures
np.random.seed(7)
temps = np.random.randint(15, 41, 28)
print("Temps:", temps)

weeks = temps.reshape(4, 7)
print("Weekly layout:")
print(weeks)

week_avg = weeks.mean(axis=1)
print("Weekly averages:", week_avg)

hottest = np.argmax(week_avg)
print(f"Hottest week: Week {hottest + 1} with avg {week_avg[hottest]:.2f}")

above_35 = np.sum(temps > 35)
print("Days above 35:", above_35)
```

```python
# Q18 - matrix multiplication not commutative
np.random.seed(42)
A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))

print("A @ B:")
print(A @ B)
print("B @ A:")
print(B @ A)
print("Are they equal?", np.array_equal(A @ B, B @ A))
# matrix multiplication is NOT commutative, A@B != B@A in general
```

```python
# Q19 - marks processing
marks = np.array([[45, 28, 67, 80],
                  [33, 90, 55, 41],
                  [72, 66, 30, 58],
                  [88, 79, 91, 84],
                  [25, 40, 38, 52]])

# replace below 35 with 35
marks[marks < 35] = 35
print("After replacing:")
print(marks)

totals = marks.sum(axis=1)
print("Totals:", totals)

topper = np.argmax(totals)
print(f"Topper: Student {topper + 1} with total {totals[topper]}")
```

```python
# Q20 - view vs copy
arr = np.arange(1, 11)

view_slice = arr[2:6]       # this is a view
copy_slice = arr[2:6].copy() # this is a copy

view_slice[0] = 999
copy_slice[0] = 999

print("Original arr:", arr)
# the view change survived - arr[2] is now 999
# the copy change did NOT affect original because copy is independent
# slices in numpy are views by default, they share memory with original
```
