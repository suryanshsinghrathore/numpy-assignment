```python
# Q2 - Price string
prices_str = " 120, 45 , 300 ,80, 95 "
parts = prices_str.split(",")
prices = []
for p in parts:
    prices.append(int(p.strip()))

total = sum(prices)
highest = max(prices)
avg = total / len(prices)

print(f"Total : {total}")
print(f"Highest : {highest}")
print(f"Average : {avg:.2f}")
```

```python
# Q4 - Sentence analysis
sentence = "Python is a very powerful programming language"
words = sentence.split()

print("Total words:", len(words))

vowels = "aeiouAEIOU"
count = 0
for w in words:
    if w[-1] in vowels:
        count = count + 1
print("Words ending with vowel:", count)

longest = words[0]
for w in words:
    if len(w) > len(longest):
        longest = w
print("Longest word:", longest)
```

```python
# Q5 - Character frequency
text = "Aba c"
text = text.lower().replace(" ", "")
freq = {}
for ch in text:
    if ch in freq:
        freq[ch] = freq[ch] + 1
    else:
        freq[ch] = 1
print(freq)
```

```python
# Q6 - Word frequency
sentence = "the cat sat on the mat and the cat sat"
words = sentence.lower().split()
word_freq = {}
for w in words:
    if w in word_freq:
        word_freq[w] = word_freq[w] + 1
    else:
        word_freq[w] = 1

sorted_words = sorted(word_freq.items(), key=lambda x: x[1], reverse=True)
print("Most frequent word:", sorted_words[0][0], "(" + str(sorted_words[0][1]) + " times)")
```

```python
# Q7 - Invert dictionary
sales = {"mon": 200, "tue": 350, "wed": 200}
inverted = {}
for day, val in sales.items():
    if val in inverted:
        inverted[val].append(day)
    else:
        inverted[val] = [day]
print(inverted)
```

```python
# Q8 - Merge marks
test1 = {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}

merged = {}
for name in test1:
    merged[name] = test1[name]

for name in test2:
    if name in merged:
        merged[name] = merged[name] + test2[name]
    else:
        merged[name] = test2[name]

print(merged)
```

```python
# Q9 - Sort employees
employees = [("Ravi", 55000), ("Sita", 72000), ("Amit", 48000)]
sorted_emp = sorted(employees, key=lambda x: x[1], reverse=True)
for name, salary in sorted_emp:
    print(f"{name} earns Rs. {salary}")
```

```python
# Q10 - Movies
movies = [
    {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
    {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
    {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
    {"title": "Movie D", "rating": 8.2, "genre": "Drama"},
]

# a) titles with rating > 8.1
print("Movies with rating > 8.1:")
for m in movies:
    if m["rating"] > 8.1:
        print(" ", m["title"])

# b) unique genres
genres = []
for m in movies:
    if m["genre"] not in genres:
        genres.append(m["genre"])
print("Genres:", genres)

# c) group by rating
rating_dict = {}
for m in movies:
    r = m["rating"]
    if r in rating_dict:
        rating_dict[r].append(m["title"])
    else:
        rating_dict[r] = [m["title"]]

print("Movies grouped by rating:")
for rating, titles in rating_dict.items():
    print(f"  Rating {rating} -> {titles}")
```
