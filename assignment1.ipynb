```python
# Q1 - Username Validator
def validate_username(username):
    username = username.strip()
    if len(username) >= 5 and username.isalnum():
        return username.lower()
    else:
        return "Invalid"

print(validate_username("  CodeMaster99 "))
print(validate_username("  Python_Dev "))
```

```python
# Q2 - Spy Code
secret_code = "charlie-bravo-alpha"
words = secret_code.split("-")
words.reverse()
result = " ".join(words).upper()
print(result)
```

```python
# Q3 - Database Cleaner
def clean_names(raw):
    names = raw.split(";")
    cleaned = []
    for n in names:
        n = " ".join(n.split())  # removes extra spaces
        n = n.title()
        cleaned.append(n)
    return ", ".join(cleaned)

raw_names = "  john doe ; jAnE sMiTh;  ALAN tUrInG "
print(clean_names(raw_names))
```

```python
# Q4 - Space Eliminator
text = "  python   is   a   very   powerful   language  "
words = text.split()
sentence = " ".join(words)
sentence = sentence[0].upper() + sentence[1:]
print(sentence)
```

```python
# Q5 - Palindrome checker
def is_palindrome(sentence):
    clean = ""
    for char in sentence.lower():
        if char.isalnum():
            clean = clean + char
    if clean == clean[::-1]:
        return True
    else:
        return False

print(is_palindrome("Was it a car or a cat I saw?"))
print(is_palindrome("Python is fun"))
```

```python
# Q6 - Hashtag extractor
post = "Loving the SUNSET! #Nature #Sunset_Vibes #beautiful"
words = post.split()
hashtags = []
for w in words:
    if w.startswith("#"):
        hashtags.append(w[1:].lower())
print(hashtags)
```

```python
# Q7 - File extensions
filenames = "report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"
files = filenames.split(", ")
extensions = []
for f in files:
    ext = f.split(".")[-1].upper()
    if ext not in extensions:
        extensions.append(ext)
print(extensions)
```

```python
# Q8 - Domain name slicer
def extract_domain(email):
    if "@" not in email:
        return "Invalid"
    domain = email.split("@")[1]
    domain = domain.split(".")[0]
    return domain.title()

print(extract_domain("developer_99@microsoft.com"))
print(extract_domain("admin_at_website.org"))
```

```python
# Q10 - Phone number formatter
def format_phone(raw):
    digits = ""
    for ch in raw:
        if ch.isdigit():
            digits = digits + ch
    if len(digits) == 10:
        return "(" + digits[:3] + ") " + digits[3:6] + "-" + digits[6:]
    else:
        return "Invalid Number"

print(format_phone(" +1 (555) 123.4567 "))
print(format_phone("555-987-6543"))
```

```python
# Q11 - SKU validator
def validate_sku(sku):
    sku = sku.strip().upper()
    if len(sku) != 8:
        return "Invalid SKU"
    if sku[3] != "-":
        return "Invalid SKU"
    first = sku[:3]
    last = sku[4:]
    if first.isalpha() and last.isdigit():
        return "Valid: " + sku
    else:
        return "Invalid SKU"

print(validate_sku("  xYz-9876 "))
print(validate_sku("AB-12345"))
print(validate_sku("A@C-1234"))
```
