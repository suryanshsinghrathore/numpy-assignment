```python
# Q2
raw_prices = " 120, 45 , 300 ,80, 95 "
price_list = []
for item in raw_prices.split(","):
    price_list.append(int(item.strip()))

t = sum(price_list)
h = max(price_list)
a = t / len(price_list)
print(f"Total : {t}")
print(f"Highest : {h}")
print(f"Average : {a:.2f}")
```

```python
# Q3
user = " Rahul_123 "
user = user.strip().lower()

valid = True
if len(user) < 5 or len(user) > 12:
    valid = False
if not user.isalnum():
    valid = False
if not user[0].isalpha():
    valid = False

if valid:
    print("Valid username")
else:
    print("Invalid username")

# testing valid case
user2 = "rahul123"
valid2 = True
if len(user2) < 5 or len(user2) > 12:
    valid2 = False
if not user2.isalnum():
    valid2 = False
if not user2[0].isalpha():
    valid2 = False

if valid2:
    print("Valid username")
else:
    print("Invalid username")
```

```python
# Q4
sent = "Learning python is really fun and exciting"
w = sent.split()
print("Total words:", len(w))

vowel_count = 0
vowels = "aeiou"
for word in w:
    if word[-1].lower() in vowels:
        vowel_count += 1
print("Words ending with vowel:", vowel_count)

big = w[0]
for word in w:
    if len(word) > len(big):
        big = word
print("Longest word:", big)
```

```python
# Q5
string = "Aba c"
string = string.lower().replace(" ", "")
char_freq = {}
for c in string:
    if c in char_freq:
        char_freq[c] += 1
    else:
        char_freq[c] = 1
print(char_freq)
```

```python
# Q6
line = "the cat sat on the mat and the cat sat"
word_list = line.lower().split()
wf = {}
for word in word_list:
    if word in wf:
        wf[word] += 1
    else:
        wf[word] = 1

# find max using sorted
sorted_wf = sorted(wf.items(), key=lambda item: item[1], reverse=True)
print("Most frequent word:", sorted_wf[0][0], "-", sorted_wf[0][1], "times")
```

```python
# Q7
sales = {"mon": 200, "tue": 350, "wed": 200}
inv = {}
for key, value in sales.items():
    if value in inv:
        inv[value].append(key)
    else:
        inv[value] = [key]
print(inv)
```

```python
# Q8
test1 = {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}

result = {}
for s in test1:
    result[s] = test1[s]
for s in test2:
    if s in result:
        result[s] = result[s] + test2[s]
    else:
        result[s] = test2[s]
print(result)
```

```python
# Q10
movies = [
    {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
    {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
    {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
    {"title": "Movie D", "rating": 8.2, "genre": "Drama"},
]

# a
print("Movies with rating > 8.1:")
for mov in movies:
    if mov["rating"] > 8.1:
        print(f"  {mov['title']}")

# b
genre_list = []
for mov in movies:
    if mov["genre"] not in genre_list:
        genre_list.append(mov["genre"])
print("Genres:", genre_list)

# c
grouped = {}
for mov in movies:
    rat = mov["rating"]
    if rat in grouped:
        grouped[rat].append(mov["title"])
    else:
        grouped[rat] = [mov["title"]]

print("Grouped by rating:")
for r in grouped:
    print(f"  {r} -> {grouped[r]}")
```
